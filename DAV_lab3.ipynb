```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set overall aesthetic style
sns.set_theme(style="whitegrid")

# 1. Create a Synthetic Dataset
np.random.seed(42)
n_samples = 100

data = {
    'Study_Hours': np.random.normal(loc=6, scale=1.5, size=n_samples),
    'Attendance_Rate': np.random.uniform(low=60, high=100, size=n_samples),
    'Previous_Score': np.random.normal(loc=70, scale=10, size=n_samples)
}

df = pd.DataFrame(data)

# Create a dependent relationship for Exam_Score
df['Exam_Score'] = (
    0.4 * df['Study_Hours'] * 10 + 
    0.3 * df['Attendance_Rate'] + 
    0.3 * df['Previous_Score'] + 
    np.random.normal(0, 3, n_samples)
)

# Categorical column for grouping
df['Grade'] = pd.cut(df['Exam_Score'], bins=[0, 70, 85, 100], labels=['C', 'B', 'A'])

print("--- Sample Dataset Head ---")
print(df.head())
```

```python
# 1. Drop Duplicate Rows
df_cleaned = df.drop_duplicates()
print("--- After Removing Duplicates ---")
print(f"Remaining rows: {len(df_cleaned)}")

# 2. Handle Missing Values for existing columns
# Fill missing 'Exam_Score' with the median score
df_cleaned['Exam_Score'] = df_cleaned['Exam_Score'].fillna(df_cleaned['Exam_Score'].median())

print("\n--- Cleaned DataFrame ---")
print(df_cleaned.head())

# Verify missing values count
print("\n--- Missing Values Re-check ---")
print(df_cleaned.isnull().sum())
```
